```python
NUM_USERS = 5_000
NUM_DEVICES = 6_000
NUM_APPLICATIONS = 100

NUM_AUTH_EVENTS = 100_000
NUM_ACCESS_EVENTS = 150_000
NUM_PRIVILEGE_EVENTS = 5_000

print(f"Users:                 {NUM_USERS:,}")
print(f"Devices:               {NUM_DEVICES:,}")
print(f"Applications:          {NUM_APPLICATIONS:,}")
print(f"Authentication events: {NUM_AUTH_EVENTS:,}")
print(f"Access events:         {NUM_ACCESS_EVENTS:,}")
print(f"Privilege events:      {NUM_PRIVILEGE_EVENTS:,}")
```

```python
FIRST_NAMES = [
    "Aarav", "Aditi", "Arjun", "Ananya", "Kabir",
    "Meera", "Rohan", "Sneha", "Rahul", "Priya",
    "Vikram", "Neha", "Karan", "Pooja", "Amit",
    "Riya", "Nikhil", "Isha", "Varun", "Kavya"
]

LAST_NAMES = [
    "Sharma", "Patel", "Mehta", "Shah", "Gupta",
    "Joshi", "Verma", "Iyer", "Reddy", "Kapoor",
    "Mishra", "Desai", "Nair", "Malhotra", "Bansal"
]

DEPARTMENTS = [
    "Finance",
    "HR",
    "Engineering",
    "IT",
    "Sales",
    "Marketing",
    "Operations",
    "Legal"
]

ROLES = [
    "Analyst",
    "Senior Analyst",
    "Manager",
    "Senior Manager",
    "Developer",
    "Administrator"
]

LOCATIONS = [
    ("Pune", "India"),
    ("Mumbai", "India"),
    ("Bengaluru", "India"),
    ("Delhi", "India"),
    ("Hyderabad", "India"),
    ("Chennai", "India"),
    ("London", "UK"),
    ("Singapore", "Singapore"),
    ("New York", "USA"),
]

OS_TYPES = [
    "Windows",
    "macOS",
    "Linux",
    "Android",
    "iOS"
]

DEVICE_TYPES = [
    "Laptop",
    "Desktop",
    "Mobile",
    "Tablet"
]

APPLICATIONS = [
    ("HR Portal", "HR", "HIGH"),
    ("Payroll Portal", "HR", "CRITICAL"),
    ("CRM", "Sales", "MEDIUM"),
    ("Jira", "Engineering", "MEDIUM"),
    ("GitHub", "Engineering", "HIGH"),
    ("AWS Console", "IT", "CRITICAL"),
    ("VPN Portal", "IT", "HIGH"),
    ("Finance Portal", "Finance", "CRITICAL"),
    ("Email", "Corporate", "MEDIUM"),
    ("Data Warehouse", "IT", "CRITICAL")
]

AUTH_METHODS = [
    "PASSWORD",
    "MFA",
    "SSO",
    "VPN"
]

AUTH_STATUSES = [
    "SUCCESS",
    "FAILED"
]

ACCESS_ACTIONS = [
    "LOGIN",
    "VIEW",
    "DOWNLOAD",
    "UPLOAD",
    "UPDATE",
    "DELETE"
]

ACCESS_STATUSES = [
    "SUCCESS",
    "DENIED"
]
```

```python
def generate_users(num_users):
    users = []

    start_date = datetime(2018, 1, 1)
    end_date = datetime(2025, 1, 1)

    for i in range(1, num_users + 1):

        user_id = f"U{i:06d}"

        first_name = random.choice(FIRST_NAMES)
        last_name = random.choice(LAST_NAMES)

        city, country = random.choice(LOCATIONS)

        join_date = (
            start_date +
            timedelta(
                days=random.randint(
                    0,
                    (end_date - start_date).days
                )
            )
        ).date()

        user = {
            "user_id": user_id,
            "employee_id": f"EMP{i:06d}",
            "first_name": first_name,
            "last_name": last_name,
            "department": random.choice(DEPARTMENTS),
            "role": random.choice(ROLES),
            "city": city,
            "country": country,
            "employment_type": random.choice(
                ["FULL_TIME", "CONTRACTOR"]
            ),
            "user_status": random.choice(
                ["ACTIVE", "ACTIVE", "ACTIVE", "INACTIVE"]
            ),
            "join_date": join_date
        }

        users.append(user)

    return users

```

```python
users = generate_users(NUM_USERS)

print(f"Generated users: {len(users):,}")
print(users[:3])
```

```python
assert len(users) == NUM_USERS
assert all(user["user_id"] for user in users)
assert len({user["user_id"] for user in users}) == NUM_USERS

print("User validation passed")
```

```python
def generate_devices(users, num_devices):

    devices = []

    for i in range(1, num_devices + 1):

        user = random.choice(users)

        device = {
            "device_id": f"D{i:06d}",
            "user_id": user["user_id"],
            "device_type": random.choice(DEVICE_TYPES),
            "os": random.choice(OS_TYPES),
            "hostname": f"DEVICE-{i:06d}",
            "is_managed": random.choice([True, True, True, False]),
            "first_seen": datetime.now() - timedelta(
                days=random.randint(1, 1000)
            )
        }

        devices.append(device)

    return devices
```

```python
devices = generate_devices(users, NUM_DEVICES)

print(f"Generated devices: {len(devices):,}")
print(devices[:3])
```

```python
user_ids = {user["user_id"] for user in users}

assert all(device["user_id"] in user_ids for device in devices)

print("Device referential integrity check passed")
```

```python
def generate_applications(num_applications):

    applications = []

    for i in range(1, num_applications + 1):

        base_app = random.choice(APPLICATIONS)

        application = {
            "application_id": f"APP{i:04d}",
            "application_name": base_app[0],
            "application_category": base_app[1],
            "sensitivity_level": base_app[2],
            "owner_department": random.choice(DEPARTMENTS)
        }

        applications.append(application)

    return applications
```

Authentication Events

```python
def generate_authentication_events(
    users,
    devices,
    num_events
):

    events = []

    user_devices = {}

    for device in devices:
        user_devices.setdefault(
            device["user_id"], []
        ).append(device)

    start_time = datetime(2026, 9, 1)

    for _ in range(num_events):

        user = random.choice(users)

        available_devices = user_devices.get(
            user["user_id"],
            []
        )

        device = (
            random.choice(available_devices)
            if available_devices
            else random.choice(devices)
        )

        timestamp = start_time + timedelta(
            minutes=random.randint(
                0,
                30 * 24 * 60
            )
        )

        status = random.choices(
            AUTH_STATUSES,
            weights=[95, 5]
        )[0]

        city, country = random.choice(LOCATIONS)

        event = {
            "event_id": str(uuid.uuid4()),
            "user_id": user["user_id"],
            "timestamp": timestamp,
            "ip_address": (
                f"{random.randint(10, 223)}."
                f"{random.randint(0, 255)}."
                f"{random.randint(0, 255)}."
                f"{random.randint(1, 254)}"
            ),
            "country": country,
            "city": city,
            "device_id": device["device_id"],
            "authentication_method": random.choice(
                AUTH_METHODS
            ),
            "login_status": status,
            "failure_reason": (
                random.choice([
                    "INVALID_PASSWORD",
                    "MFA_FAILED",
                    "ACCOUNT_LOCKED"
                ])
                if status == "FAILED"
                else None
            )
        }

        events.append(event)

    return events
```

```python
auth_events = generate_authentication_events(
    users,
    devices,
    NUM_AUTH_EVENTS
)

print(f"Authentication events: {len(auth_events):,}")
print(auth_events[:2])
```

ACCESS EVENTS

```python
def generate_access_events(
    users,
    devices,
    applications,
    num_events
):

    events = []

    start_time = datetime(2026, 9, 1)

    for _ in range(num_events):

        user = random.choice(users)
        device = random.choice(devices)
        application = random.choice(applications)

        timestamp = start_time + timedelta(
            minutes=random.randint(
                0,
                30 * 24 * 60
            )
        )

        action = random.choice(ACCESS_ACTIONS)

        status = random.choices(
            ACCESS_STATUSES,
            weights=[97, 3]
        )[0]

        events.append({
            "event_id": str(uuid.uuid4()),
            "user_id": user["user_id"],
            "timestamp": timestamp,
            "application_id": application["application_id"],
            "resource": (
                f"{application['application_name']}_"
                f"RESOURCE"
            ),
            "action": action,
            "device_id": device["device_id"],
            "ip_address": (
                f"{random.randint(10, 223)}."
                f"{random.randint(0, 255)}."
                f"{random.randint(0, 255)}."
                f"{random.randint(1, 254)}"
            ),
            "access_status": status
        })

    return events
```

```python
access_events = generate_access_events(
    users,
    devices,
    applications,
    NUM_ACCESS_EVENTS
)

print(f"Access events: {len(access_events):,}")
print(access_events[:2])
```

Priviledge Events

```python
def generate_privilege_events(
    users,
    num_events
):

    events = []

    roles = [
        "Analyst",
        "Senior Analyst",
        "Manager",
        "Administrator"
    ]

    start_time = datetime(2026, 9, 1)

    for _ in range(num_events):

        user = random.choice(users)

        old_role = random.choice(roles)
        new_role = random.choice(roles)

        events.append({
            "event_id": str(uuid.uuid4()),
            "user_id": user["user_id"],
            "timestamp": start_time + timedelta(
                minutes=random.randint(
                    0,
                    30 * 24 * 60
                )
            ),
            "old_role": old_role,
            "new_role": new_role,
            "changed_by": random.choice(users)["user_id"],
            "approval_status": random.choice([
                "APPROVED",
                "APPROVED",
                "PENDING",
                "REJECTED"
            ])
        })

    return events
```

```python
privilege_events = generate_privilege_events(
    users,
    NUM_PRIVILEGE_EVENTS
)

print(f"Privilege events: {len(privilege_events):,}")
print(privilege_events[:2])
```
